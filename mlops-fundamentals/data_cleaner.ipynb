```python
import pandas as pd
import json
```

```python
# Step 1: Load data
df = pd.read_csv("mock_data.csv")
print(df.head())
```

```python
# Step 2: Fill missing values for age and salary
df["age"].fillna(df["age"].median(), inplace=True)
df["salary"].fillna(df["salary"].median(), inplace=True)
```

```python
df.fillna({"department":"Unknown"}, inplace=True)
```

```python
# Step 4: Convert JSON strings in "profile" column
df["profile"] = df["profile"].apply(lambda x: json.loads(x) if pd.notnull(x) else {})
```

```python
# Step 5: Extract nested fields
df["address"] = df["profile"].apply(lambda x: x.get("address", None))
df["phone"] = df["profile"].apply(lambda x: x.get("phone", None))
df["email"] = df["profile"].apply(lambda x: x.get("email", None))
```

```python
df.drop(columns=["profile"], inplace=True)
```

```python
# Save cleaned data
df.to_csv("cleaned_data.csv", index=False)
print("Data cleaned and saved to 'cleaned_data.csv'.")
```
