```python
import pandas as pd
```

```python
### Step 1: Load cleaned data
df = pd.read_csv("cleaned_data.csv")
```

```python
### Step 2: Add address_length column
df["address_length"] = df["address"].apply(lambda x: len(str(x)))
```

```python
### Step 3: Create salary_category column
bins = [0, 50000, 70000, 100000]
labels = ["low", "medium", "high"]
df["salary_category"] = pd.cut(df["salary"], bins=bins, labels=labels, include_lowest=True)
```

```python
### Step 4: Group by department and calculate averages
summary_report = df.groupby("department").agg({
    "salary": "mean",
    "age": "mean"
}).reset_index()
print(summary_report)
```

```python
### Save transformed data
df.to_csv("transformed_data.csv", index=False)
print("Transformed data saved to 'transformed_data.csv'")
```
